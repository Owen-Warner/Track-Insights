```python
#This is set for 2025
#Boys
meter100B = 11.01
meter200B = 22.21
meter400B = 49.07
meter800B = 115.21
meter1600B = 256.7
meter3200B = 550.91
discusB = 1936
shotPutB = 676
longJumpB = 263.5
highJumpB = 77
poleVaultB = 174
#Girls
meter100G = 12.24
meter200G = 25.34
meter400G = 57.44
meter800G = 134.68
meter1600G = 298.87
meter3200G = 563.63
discusG = 1,527
shotPutG = 501.25
longJumpG = 216.25
highJumpG = 64
poleVaultG = 135
```

```python
fieldList = ["Long Jump","High Jump","Shot Put","Pole Vault","Discus"]
```

```python
def stateStandard(filt2, event, start, gender):
    filt2 = filt2.sort_values(by='result2')
    if event in fieldList:
        filt2 = filt2.sort_values(by='result2', ascending=False)
    if gender == "Boys":
        if event == "100 Meters":
            val = filt2.iloc[start]
            while val.iloc[8] <= meter100B:
                start+=1
                val = filt2.iloc[start]
        elif event == "200 Meters":
            val = filt2.iloc[start]
            while val.iloc[8] <= meter200B:
                start+=1
                val = filt2.iloc[start]
        elif event == "400 Meters":
            val = filt2.iloc[start]
            while val.iloc[8] <= meter400B:
                start+=1
                val = filt2.iloc[start]
        elif event == "800 Meters":
            val = filt2.iloc[start]
            while val.iloc[8] <= meter800B:
                start+=1
                val = filt2.iloc[start]
        elif event == "1600 Meters":
            val = filt2.iloc[start]
            while val.iloc[8] <= meter1600B:
                start+=1
                val = filt2.iloc[start]
        elif event == "3200 Meters":
            val = filt2.iloc[start]
            while val.iloc[8] <= meter3200B:
                start+=1
                val = filt2.iloc[start]
        elif event == "Discus":
            val = filt2.iloc[start]
            while val.iloc[8] >= discusB:
                start+=1
                val = filt2.iloc[start]
        elif event == "Shot Put":
            val = filt2.iloc[start]
            while val.iloc[8] >= shotPutB:
                start+=1
                val = filt2.iloc[start]
        elif event == "Long Jump":
            val = filt2.iloc[start]
            while val.iloc[8] >= longJumpB:
                start+=1
                val = filt2.iloc[start]
        elif event == "High Jump":
            val = filt2.iloc[start]
            while val.iloc[8] >= highJumpB:
                start+=1
                val = filt2.iloc[start]
        elif event == "Pole Vault":
            val = filt2.iloc[start]
            while val.iloc[8] >= poleVaultB:
                start+=1
                val = filt2.iloc[start]
    elif gender == "Girls":
        if event == "100 Meters":
            val = filt2.iloc[start]
            while val.iloc[8] <= meter100G:
                start+=1
                val = filt2.iloc[start]
        elif event == "200 Meters":
            val = filt2.iloc[start]
            while val.iloc[8] <= meter200G:
                start+=1
                val = filt2.iloc[start]
        elif event == "400 Meters":
            val = filt2.iloc[start]
            while val.iloc[8] <= meter400G:
                start+=1
                val = filt2.iloc[start]
        elif event == "800 Meters":
            val = filt2.iloc[start]
            while val.iloc[8] <= meter800G:
                start+=1
                val = filt2.iloc[start]
        elif event == "1600 Meters":
            val = filt2.iloc[start]
            while val.iloc[8] <= meter1600G:
                start+=1
                val = filt2.iloc[start]
        elif event == "3200 Meters":
            val = filt2.iloc[start]
            while val.iloc[8] <= meter3200G:
                start+=1
                val = filt2.iloc[start]
        elif event == "Discus":
            val = filt2.iloc[start]
            while val.iloc[8] >= discusG:
                start+=1
                val = filt2.iloc[start]
        elif event == "Shot Put":
            val = filt2.iloc[start]
            while val.iloc[8] >= shotPutG:
                start+=1
                val = filt2.iloc[start]
        elif event == "Long Jump":
            val = filt2.iloc[start]
            while val.iloc[8] >= longJumpG:
                start+=1
                val = filt2.iloc[start]
        elif event == "High Jump":
            val = filt2.iloc[start]
            while val.iloc[8] >= highJumpG:
                start+=1
                val = filt2.iloc[start]
        elif event == "Pole Vault":
            val = filt2.iloc[start]
            while val.iloc[8] >= poleVaultG:
                start+=1
                val = filt2.iloc[start]
    start+=1
    return start
```

```python
#State Standards 2024: https://www.scribd.com/document/734190032/2024-IHSAA-State-Standards-3-Participant-Standard

podium_df = regional_df[(regional_df["place"] <= 3)]
events = db.get_event()
genders = db.get_gender()
count = 0
for g in genders["gender"]:
    for x in events["event"]:
        filt = other_df[(other_df['event'] == x) & ((other_df['result2'] < 9999) | (other_df['result2'] > 0)) & (other_df["gender"] == g)]
        filt = filt.sort_values(by='result2')
        if filt.empty:
            blank = 1
        else:
            headsize = 3
            filt2 = filt.head(len(filt)) #I do not know why it needs this, but filt does not work
            filt2.sort_values(by='first')
            filt2 = filt2.sort_values(by='result2')
            if x in fieldList:
                filt2.sort_values(by='first')
                filt2 = filt2.sort_values(by='result2', ascending=False)
            index_last = 2
            old = filt2.iloc[2]
            new = filt2.iloc[index_last+1]
            while old.iloc[8] == new.iloc[8]: #Something strange here is adding lots of rows with wierd new column, no old
                count += 1
                headsize += 1
                index_last += 1
                old = filt2.iloc[index_last]
                new = filt2.iloc[index_last+1]
            if x == "Discus":
                #print(filt2)
                val = filt2.iloc[headsize]
                print(val.iloc[8])
            headsize = stateStandard(filt2, x, headsize, g)-1 #New
            podium_df = pd.concat([podium_df, filt2.head(headsize)], ignore_index=True) #New
    #use head iloc instead of filt iloc
```
